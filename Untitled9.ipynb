```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from datetime import date as dt_class

# ==========================================
#  HELPER FUNCTIONS
# ==========================================
@st.cache_data
def load_data(uploaded_file):
    if uploaded_file is not None:
        uploaded_file.seek(0)
        return pd.read_csv(uploaded_file)
    return None

def preprocess_data(df):
    data = df.copy()

    # 1. Standardize Date
    if 'date' in data.columns:
        data['date'] = pd.to_datetime(data['date'])
        data['month'] = data['date'].dt.month
        data['day_of_week'] = data['date'].dt.day_name()
        data['is_weekend'] = data['date'].dt.dayofweek.apply(lambda x: 1 if x >= 5 else 0)

    target_col = "actual_units_sold"

    # Drop IDs and targets from features
    drop_cols_training = ["id", "predicted_demand_pct", "recommended_units", target_col]

    return data, drop_cols_training, target_col

def build_model_pipeline(model_type, numeric_cols, cat_cols, params):
    numeric_transformer = StandardScaler()
    categorical_transformer = OneHotEncoder(handle_unknown='ignore')

    preprocessor = ColumnTransformer(
        transformers=[
            ('num', numeric_transformer, numeric_cols),
            ('cat', categorical_transformer, cat_cols)
        ])

    if model_type == "Linear Regression":
        clf = LinearRegression()
    elif model_type == "Random Forest":
        clf = RandomForestRegressor(n_estimators=params['rf_n_estimators'],
                                    max_depth=params['rf_max_depth'],
                                    random_state=42)
    elif model_type == "XGBoost":
        clf = XGBRegressor(n_estimators=params['xgb_n_estimators'],
                           learning_rate=params['xgb_lr'],
                           max_depth=params['xgb_max_depth'],
                           random_state=42)
    else:
        clf = LinearRegression()

    return Pipeline(steps=[('preprocessor', preprocessor), ('model', clf)])

# ==========================================
#  MAIN APP UI
# ==========================================
st.set_page_config(page_title="CodeNova Forecaster", layout="wide")
st.title("🍎 CodeNova: Food Demand AI")

# --- SIDEBAR ---
st.sidebar.header("1. Upload Data")
uploaded_file = st.sidebar.file_uploader("Upload CSV", type=["csv"])

# INITIALIZE VARIABLES
raw_df = None
selected_foods_train = []

if uploaded_file:
    raw_df = load_data(uploaded_file)

if raw_df is not None:
    st.sidebar.header("2. Training Filters")
    if 'food_name' in raw_df.columns:
        unique_foods = sorted(raw_df['food_name'].astype(str).unique())
        selected_foods_train = st.sidebar.multiselect(
            "Select Food(s) to Train On (Leave empty for ALL)",
            options=unique_foods,
            default=[]
        )
    else:
        st.sidebar.warning("Column 'food_name' not found.")

st.sidebar.header("3. Hyperparameters")
rf_n = st.sidebar.slider("RF Trees", 50, 300, 100)
xgb_lr = st.sidebar.number_input("XGB Learning Rate", 0.01, 0.5, 0.1)
params = {'rf_n_estimators': rf_n, 'rf_max_depth': 10, 'xgb_n_estimators': 100, 'xgb_lr': xgb_lr, 'xgb_max_depth': 6}

# --- MAIN LOGIC ---
if raw_df is not None:
    # Preprocess
    processed_df, drop_cols, target = preprocess_data(raw_df)

    # FILTERING
    if selected_foods_train:
        filtered_df = processed_df[processed_df['food_name'].isin(selected_foods_train)].copy()
        st.info(f"ℹ️ Training Mode: specific foods ({', '.join(selected_foods_train)}) | Records: {len(filtered_df)}")
    else:
        filtered_df = processed_df.copy()
        st.info(f"ℹ️ Training Mode: All Foods | Records: {len(filtered_df)}")

    # Define Features
    feature_cols = [c for c in filtered_df.columns if c not in drop_cols and c != 'date' and c != target]
    numeric_features = filtered_df[feature_cols].select_dtypes(include=['int64', 'float64']).columns.tolist()
    categorical_features = filtered_df[feature_cols].select_dtypes(include=['object']).columns.tolist()

    # TABS
    tab1, tab2, tab3 = st.tabs(["📊 Data View", "🤖 Training", "🔮 Prediction"])

    with tab1:
        st.dataframe(filtered_df.head())

    with tab2:
        st.subheader("Train Your Model")
        if st.button("Start Training"):
            if len(filtered_df) < 5:
                st.error("Not enough data to train! Please select more foods.")
            else:
                X = filtered_df[numeric_features + categorical_features]
                y = filtered_df[target]

                # Split
                X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

                # Train
                model = build_model_pipeline("XGBoost", numeric_features, categorical_features, params)
                model.fit(X_train, y_train)

                # Save
                st.session_state['model'] = model
                st.session_state['numeric_features'] = numeric_features
                st.session_state['categorical_features'] = categorical_features
                st.session_state['reference_df'] = filtered_df

                preds = model.predict(X_test)
                mae = mean_absolute_error(y_test, preds)
                st.success(f"Training Complete! MAE: {mae:.2f}")

                # Chart
                res_df = pd.DataFrame({'Actual': y_test, 'Predicted': preds})
                st.plotly_chart(px.scatter(res_df, x='Actual', y='Predicted', title="Actual vs Predicted"), use_container_width=True)

    with tab3:
        st.subheader("Custom Prediction")
        if 'model' not in st.session_state:
            st.warning("⚠️ Please train the model first.")
        else:
            col_a, col_b = st.columns(2)

            with col_a:
                st.markdown("#### Input Scenario")
                available_foods = selected_foods_train if selected_foods_train else sorted(raw_df['food_name'].unique())
                food_input = st.selectbox("Select Food Item", available_foods)
                date_input = st.date_input("Select Date", value=dt_class.today())

                ref_df = st.session_state['reference_df']
                food_stats = ref_df[ref_df['food_name'] == food_input]

                default_price = 50.0
                if not food_stats.empty:
                    default_price = float(food_stats['price_per_unit'].mean())

                price_input = st.number_input("Price per Unit (₹)", value=default_price)
                discount_input = st.number_input("Discount (%)", value=0.0)

            with col_b:
                st.markdown("#### Result")
                if st.button("Predict Demand"):
                    if food_stats.empty:
                        base_row = ref_df.iloc[[0]].copy()
                    else:
                        base_row = food_stats.iloc[[-1]].copy()

                    # --- FIXED DATE LOGIC HERE ---
                    # We treat the date as a single object, not a pandas Series
                    d_val = pd.to_datetime(date_input)

                    base_row['food_name'] = food_input
                    base_row['date'] = d_val
                    base_row['month'] = d_val.month
                    base_row['day_of_week'] = d_val.day_name()
                    base_row['is_weekend'] = 1 if d_val.weekday() >= 5 else 0

                    base_row['price_per_unit'] = price_input
                    base_row['discount_pct'] = discount_input

                    model_cols = st.session_state['numeric_features'] + st.session_state['categorical_features']
                    input_row = base_row[model_cols]

                    model = st.session_state['model']
                    pred = model.predict(input_row)[0]

                    st.metric(label="Predicted Sales", value=f"{int(pred)}")
                    if pred > 100: st.success("High Demand!")
                    else: st.info("Standard Demand.")
else:
    st.info("Waiting for upload...")
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
from google.colab import files  # Import library to download files

# --- CONFIGURATION ---
num_records = 500
start_date = datetime(2024, 1, 1) # Future data (2024)
foods = ['Masala Dosa', 'Idli Vada', 'Rava Idli', 'Mysore Masala Dosa', 'Rice Bath', 'Vada Pav', 'Filter Coffee']
regions = ['Indiranagar', 'Koramangala', 'Jayanagar', 'Whitefield', 'Malleshwaram']

np.random.seed(99)

# --- GENERATE DATA ---
data = {
    'id': range(1001, 1001 + num_records),
    'date': [start_date + timedelta(days=np.random.randint(0, 90)) for _ in range(num_records)],
    'food_name': np.random.choice(foods, num_records),
    'region': np.random.choice(regions, num_records),
    'price_per_unit': np.random.uniform(35, 160, num_records).round(2),
    'discount_pct': np.random.uniform(0, 15, num_records).round(1),
    'temperature_c': np.random.uniform(18, 32, num_records).round(1),
    'is_holiday': np.random.choice([0, 1], num_records, p=[0.95, 0.05]),
    'competitor_density': np.random.randint(1, 10, num_records),
    # Dummy columns for compatibility
    'predicted_demand_pct': np.random.uniform(0.5, 1.5, num_records).round(2),
    'recommended_units': np.random.randint(50, 200, num_records),
}

df_test = pd.DataFrame(data)

# Logic for Target (Actual Sales)
df_test['actual_units_sold'] = (
    160
    - (df_test['price_per_unit'] * 0.6)
    + (df_test['discount_pct'] * 1.5)
    + (df_test['is_holiday'] * 60)
    + np.random.normal(0, 12, num_records)
).astype(int)

df_test['actual_units_sold'] = df_test['actual_units_sold'].apply(lambda x: max(10, x))
df_test = df_test.sort_values('date')

# --- SAVE & DOWNLOAD ---
file_name = 'food_demand_testing_2024.csv'
df_test.to_csv(file_name, index=False)

print(f"✅ Generated {file_name}")
print("⬇️ Downloading file to your computer...")
files.download(file_name)
```
